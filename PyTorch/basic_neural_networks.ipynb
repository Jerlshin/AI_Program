### Import

```python
import torch
import torch.nn as nn
```

### Model

```python
model = nn.Sequential(
    nn.Linear(784, 10),
    nn.ReLU(),
    nn.Linear(10, 10), # Output layer with 10 classes
)
```

```python
model
```

### Data

```python
x = torch.randn(1, 784) # 1 row, 784 features --> 1 sample
```

```python
x.shape # Vector
```

```python
x.type  # Tensor
```

```python
# Forward pass
out = model(x) # Output from the model
```

```python
out # logits - raw scores from the model's layer
```

```python
out.shape
```

```python
softmax = nn.Softmax(dim=0) # Columns instead of rows
probs = softmax(out)
probs
```

```python
softmax = nn.Softmax(dim=1)
probs = softmax(out)
probs # 10 classes
```

```python
output_class = torch.max(probs)
```

```python
output_class
```

```python
output_class = torch.argmax(probs) # to get the index
```

```python
output_class # the input image is 8 # y: output from the model
```

### Training

```python
# loss calculation
criterion = nn.CrossEntropyLoss() # loss function
target = torch.tensor([8]) # Target class # Y: actual label

loss = criterion(out, target)
loss
```

```python
model.parameters()
```

* Optimizers
* Schedulers

```python
optimizer = torch.optim.Adam(model.parameters(), lr = 0.01) # alpha
```

```python
n_epochs = 10
```

```python
target = torch.tensor([2])
```

```python
for epoch in range(n_epochs):
    y_pred = model(x) # Forward pass
    y_probs = nn.Softmax(dim=1)(y_pred) # Softmax
    
    print('epoch: ', epoch,' loss: ', loss.item())

    loss = criterion(y_pred, target) # Loss calculation
    optimizer.zero_grad() # Zero gradients
    loss.backward() # Backpropagation
    optimizer.step() # Update weights
```
